```python
df = pd.read_excel('/content/drive/MyDrive/kyoto.xlsx', header=4)

print(df.head())
```

```python
df.columns = ['year', 'households', 'population', 'male', 'female',
              'x1', 'people_per_house', 'density', 'area', 'note']

print(df.head())
```

```python
df = df.replace(999999999, pd.NA)
df = df[['year', 'population', 'male', 'female']]
print(df.head())
```

```python
print(df.tail(20))
```

```python
df_year = df[df['year'].astype(str).str.contains('令和', na=False)].copy()

print(df_year)
df_year['year_number'] = df_year['year'].str.extract(r'\((\d{4})\)')
print(df_year[['year', 'year_number', 'population']])
import matplotlib.pyplot as plt
plt.plot(df_year['year_number'],
         df_year['population'],
         marker='o')
plt.title('Kyoto Population Change')
plt.xlabel('Year')
plt.ylabel('Population')
plt.savefig("population_chart.png")
plt.show()
```

```python
plt.plot(df_year['year_number'], df_year['male'], label='Male')
plt.plot(df_year['year_number'], df_year['female'], label='Female')
marker='o'
plt.grid(True)
plt.legend()
plt.savefig(".png")
plt.show()
```

```python
df_year['change'] = df_year['population'].diff()


colors = ['blue' if x < 0 else 'black' for x in df_year['change']]


import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(df_year['year_number'],
        df_year['change'],
        color=colors)

plt.title('Annual Population Change in Kyoto')
plt.xlabel('Year')
plt.ylabel('Population Change')

plt.grid(True)
plt.savefig("population_change_bar_chart.png")
plt.show()
```

```python
import matplotlib.pyplot as plt


male = df_year.iloc[-1]['male']
female = df_year.iloc[-1]['female']


sizes = [male, female]


labels = ['Male', 'Female']


plt.figure(figsize=(6,6))

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%'
)

plt.title('Population Composition by Gender')


plt.savefig("gender_composition_2026.png")

plt.show()
```
